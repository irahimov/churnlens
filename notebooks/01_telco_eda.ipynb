```python
mean_monthly = df['MonthlyCharges'].mean()
services = ['OnlineSecurity', 'OnlineBackup', 'TechSupport', 'InternetService']
high_paying_df = df.loc[df['MonthlyCharges'] > mean_monthly]
result = high_paying_df.groupby(services, observed=False).agg(
    Musteri_Sayi=('Churn', 'count'),
    Churn_Faizi=('Churn', lambda x: (x == 'Yes').mean() * 100),
    Orta_Odenis=('MonthlyCharges', 'mean')
).reset_index()
result.sort_values(by='Churn_Faizi', ascending=False)
```

## Nəticələr (Faza 1)

**Churn balansı.** Data dengesizdir: müştərilərin təxminən 73%-i churn etməyib (No), 26%-i churn edib (Yes). Yalnız accuracy ilə qiymətləndirmək yanıltıcı olar.

**Tenure təsiri.** Churn tenure ilə kəskin azalır: 0–10 aylıq aralıqda ~49%, 60–70 aylıqda ~9%. Ən böyük sıçrayış ilk 10–20 ay arasındadır. Bu, ilk aylarda müştəri ilə əlaqənin ən çox fərq yaratdığını göstərir. Qrupların müştəri sayı da yoxlanmalıdır, çünki kiçik qrupda faiz dəyişkən olur.

**TotalCharges.** Sütun mətn kimi gəlir, çünki 11 sətirdə boşluq (`" "`) var. Hamısının `tenure = 0`-dır, yəni hələ ödəniş dövrü tamamlanmayan yeni müştərilərdir. Bu sətirlər `fillna(0)` ilə doldurulub, `new_customer = (tenure == 0)` flag-i ilə ayrıca işarələnəcək. Bu 11 müştərinin hamısı churn etməyib (No).

**Qərar.** Modeldə tenure, MonthlyCharges və TotalCharges (rəqəm kimi) istifadə olunacaq. TotalCharges-ın digər sütunlarla əlaqəsi Faza 2-də yoxlanacaq. Churn ilə bağlı sütunlar (Churn Score, CLTV və s.) bu datada yoxdur, amma hər sütun üçün leakage yoxlaması Faza 2-də ediləcək.
